```python
# Problem 6: Combining Filtering and Sorting

def find_top_student_in_major(df, major_name):
    if major_name not in df['Major'].unique():
        print(f"Warning: Major '{major_name}' not found in the dataset.")
        return None

    filtered = df[df['Major'] == major_name]
    sorted_df = filtered.sort_values(by='Score', ascending=False)
    top_student = sorted_df.iloc[0]
    return top_student[['Name', 'Score']]

print("Top in Computer Science:")
print(find_top_student_in_major(student_df, 'Computer Science'))

print("\nTop in Physics:")
print(find_top_student_in_major(student_df, 'Physics'))

print("\nTop in Chemistry:")
print(find_top_student_in_major(student_df, 'Chemistry'))
```

## Part 2: Processing tabular data using Pandas

```python
# Load datasets
stroke_df = pd.read_csv('/content/drive/MyDrive/AIHC5020/stroke/healthcare-dataset-stroke-data.csv')
location_codes_df = pd.read_csv('/content/drive/MyDrive/AIHC5020/stroke/location_codes.csv')
```

```python
# Problem 7: Applying row-level operations

def check_cardiovascular(row):
    if row['hypertension'] == 1 or row['heart_disease'] == 1:
        return 1
    return 0

stroke_df['cardiovascular_disease'] = stroke_df.apply(check_cardiovascular, axis=1)

print("--- Verification of cardiovascular_disease column ---")
display(stroke_df[['hypertension', 'heart_disease', 'cardiovascular_disease']].head())
print("\nValue Counts:")
print(stroke_df['cardiovascular_disease'].value_counts())
```

```python
# Problem 8: Exploratory visualizations

def plot_histograms_by_category(df, continuous_col, categorical_col):
    # Edge case 2: Check existence
    if continuous_col not in df.columns or categorical_col not in df.columns:
        print(f"Error: Column(s) not found in DataFrame.")
        return None

    # Edge case 1: Heuristic check for types
    if df[continuous_col].nunique() < 10:
        print(f"Error: {continuous_col} appears to be categorical (fewer than 10 unique values).")
        return None
    if df[categorical_col].nunique() > 10:
        print(f"Error: {categorical_col} appears to be continuous (more than 10 unique values).")
        return None

    fig, ax = plt.subplots(figsize=(10, 6))
    categories = df[categorical_col].unique()

    for cat in categories:
        subset = df[df[categorical_col] == cat]
        ax.hist(subset[continuous_col], alpha=0.6, label=f'{categorical_col} = {cat}')

    ax.set_title(f'Distribution of {continuous_col} by {categorical_col}')
    ax.set_xlabel(continuous_col)
    ax.set_ylabel('Frequency')
    ax.legend()
    plt.show()

# Tests
plot_histograms_by_category(stroke_df, 'age', 'stroke')
plot_histograms_by_category(stroke_df, 'bmi', 'gender')
plot_histograms_by_category(stroke_df, 'stroke', 'age') # Edge case test
```

```python
# Problem 9: Group-wise Statistics

def calculate_stroke_prevalence(df, group_by_cols):
    # Edge case: existence
    for col in group_by_cols:
        if col not in df.columns:
            print(f"Error: Column '{col}' not found.")
            return None
        # Edge case: heuristic categorical check
        if df[col].nunique() > 15:
             print(f"Error: Column '{col}' seems non-categorical (high cardinality).")
             return None

    grouped = df.groupby(group_by_cols)
    sizes = grouped.size()
    strokes = grouped['stroke'].sum()

    prevalence = (strokes / sizes).to_frame(name='stroke_prevalence').reset_index()
    return prevalence

# Tests
print("Prevalence by Smoking Status:")
display(calculate_stroke_prevalence(stroke_df, ['smoking_status']))

print("\nPrevalence by Marriage and Work:")
display(calculate_stroke_prevalence(stroke_df, ['ever_married', 'work_type']))

print("\nCase: Age (Non-categorical):")
calculate_stroke_prevalence(stroke_df, ['age'])

print("\nCase: Capitalization Error:")
calculate_stroke_prevalence(stroke_df, ['residence_type'])
```

```python
# Problem 10: Building a Preprocessing Pipeline

def preprocess_for_analysis(df, cols_to_impute, cols_to_scale, cols_to_encode, cols_to_drop):
    new_df = df.copy()

    # a. Drop
    new_df = new_df.drop(columns=cols_to_drop)

    # b. Impute (Mean)
    for col in cols_to_impute:
        new_df[col] = new_df[col].fillna(new_df[col].mean())

    # c. Scale
    for col in cols_to_scale:
        mean = new_df[col].mean()
        std = new_df[col].std()
        new_df[f'{col}_scaled'] = (new_df[col] - mean) / std
        new_df = new_df.drop(columns=[col])

    # d. One-hot encode
    new_df = pd.get_dummies(new_df, columns=cols_to_encode, drop_first=True)

    return new_df

# Define lists
impute_list = ['bmi']
scale_list = ['age', 'avg_glucose_level', 'bmi']
encode_list = ['gender', 'ever_married', 'work_type', 'Residence_type', 'smoking_status']
drop_list = ['id', 'location_code']

processed_df = preprocess_for_analysis(stroke_df, impute_list, scale_list, encode_list, drop_list)
display(processed_df.head())
processed_df.info()
```

```python
# Problem 11: Joins with Pandas

merged_df = stroke_df.merge(location_codes_df, on='location_code', how='left')
print("--- Merged DataFrame Head ---")
display(merged_df.head())
```

```python
# Problem 12: Introduction to SQL with DuckDB
import duckdb

sql_query = """
SELECT *
FROM stroke_df AS l
LEFT JOIN location_codes_df AS r
ON l.location_code = r.location_code
WHERE r.state = 'NY'
"""

sql_result_df = duckdb.query(sql_query).to_df()

print("--- SQL Result Head ---")
display(sql_result_df.head())

print("\nUnique states in result:")
print(sql_result_df['state'].unique())
```
